```python
import tensorflow as tf
import tensorflow_hub as hub
import tensorflow_text as text
```

```python
preprocessor_path = "./bert_preprocess"
encoder_path = "./bert_encoder"
```

```python
preprocessor = hub.KerasLayer(preprocessor_path)
encoder = hub.KerasLayer(encoder_path)
```

```python
text_input = tf.constant([
    "I love running BERT offline on my local computer!",
    "TensorFlow Hub models are powerful."
])
```

```python
encoder_inputs = preprocessor(text_input)
```

```python
encoder_inputs.keys()
```

```python
encoder_inputs['input_mask']
```

```python
outputs = encoder(encoder_inputs)
```

```python
outputs.keys()
```

```python
pooled_output = outputs["pooled_output"]       # Sentence-level embedding
sequence_output = outputs["sequence_output"]   # Token-level embedding
```

```python
print("✅ Pooled output shape:", pooled_output.shape)
print("✅ Sequence output shape:", sequence_output.shape)
```
