```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, Dense, SimpleRNN, LSTM, GRU, Bidirectional

# Parameters
max_features = 10000   # Vocabulary size
maxlen = 200           # Max length of sequences
embedding_dim = 128

# Load IMDB dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=max_features)

# Pad sequences
x_train = pad_sequences(x_train, maxlen=maxlen)
x_test = pad_sequences(x_test, maxlen=maxlen)
```

```python
model = Sequential()
model.add(Embedding(max_features, embedding_dim))
model.build(input_shape=(None, maxlen))
model.add(SimpleRNN(64))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2)
```

```python
model_rnn = Sequential()
model_rnn.add(Embedding(max_features, embedding_dim))
model_rnn.build(input_shape=(None,maxlen))
model_rnn.add(Bidirectional(SimpleRNN(64)))
model_rnn.add(Dense(1, activation='sigmoid'))

model_rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model_rnn.summary()
```

```python
model_rnn.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2)
```

```python
model_lstm = Sequential()
model_lstm.add(Embedding(max_features, embedding_dim))
model_lstm.build(input_shape=(None,maxlen))
model_lstm.add(Bidirectional(LSTM(64)))
model_lstm.add(Dense(1, activation='sigmoid'))

model_lstm.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model_lstm.summary()
```

```python
model_lstm.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2)
```

```python
model_gru = Sequential()
model_gru.add(Embedding(max_features, embedding_dim))
model_gru.build(input_shape=(None,maxlen))
model_gru.add(Bidirectional(GRU(64)))
model_gru.add(Dense(1, activation='sigmoid'))

model_gru.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

```

```python
model_gru.summary()
```

```python
model_gru.fit(x_train, y_train, epochs=3, batch_size=64, validation_split=0.2)
```
