```python
x= df.iloc[:,0:2]
y= df.iloc[:,-1]
```

```python
x.head()
```

```python
from sklearn.linear_model import Perceptron
p= Perceptron(max_iter=10000, eta0=1.0, random_state=42, tol=1e-3)
```

```python
p.fit(x,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.values, y.values , clf=p , legend=2)
```
