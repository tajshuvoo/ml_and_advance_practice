```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id','Unnamed: 32'] , inplace=True)
```

```python
df.head()
```

```python
x_train, x_test , y_train , y_test = train_test_split(df.iloc[:,1:], df.iloc[:,0], test_size=0.2)
```

```python
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
label = LabelEncoder()
y_train = label.fit_transform(y_train)
y_test = label.transform(y_test)
```

Numpy to PyTorch tensor 

```python
x_train = torch.from_numpy(x_train)
x_test = torch.from_numpy(x_test)
y_train = torch.from_numpy(y_train)
y_test = torch.from_numpy(y_test)
```

Defining the MODEL

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):
    
    def __init__(self, features , labels):
        
        self.features = features
        self.labels = labels
        
    def __len__(self):
        return self.features.shape[0]
    
    def __getitem__(self, index):
        return self.features[index] , self.labels[index]
```

```python
train_dataset = CustomDataset(x_train, y_train)
test_dataset = CustomDataset(x_test , y_test)
```

```python
train_dataset[10]
```

```python
train_loader = DataLoader(train_dataset , batch_size=32 , shuffle=True)
test_loader = DataLoader(test_dataset , batch_size=32 , shuffle=True)
```

```python
import torch
import torch.nn as nn
```

```python
class MySimpleNN(nn.Module):
    
    def __init__(self , num_features):
        
        super().__init__()
        
        self.linear = nn.Linear(num_features , 1)
        self.sigmoid = nn.Sigmoid()
        
        
    def forward(self, features):
        
        out = self.linear(features)
        out = self.sigmoid(out)
        
        return out
    

    
```

```python
learning_rate = 0.1
epochs = 25
```

```python
model = MySimpleNN(x_train.shape[1])
optimizer = torch.optim.SGD(model.parameters() ,lr= learning_rate)
loss_function= nn.BCELoss()
```

Training PipeLine

```python

for epoch in range(epochs):
    
    for batch_features , batch_labels in train_loader:
        # Convert to float32 for compatibility with model
        batch_features = batch_features.float()
        batch_labels = batch_labels.float()
        
        #forward pass
        y_pred = model(batch_features)
        
        #loss calculate
        loss = loss_function(y_pred, batch_labels.view(-1,1))
        
        #zero grad for next epoch not needed accumulate for grad
        # It is better to make it zero before calculation than after 
        optimizer.zero_grad()
        #backward pass 
        loss.backward()
        #parameter update
        optimizer.step()
        
        print(loss.item())
```

```python
model.linear.weight, model.linear.bias
```

```python
#model evaluation
model.eval()
accuracy_list= []

with torch.no_grad():
    for batch_features , batch_labels in test_loader:
        # Convert batch_features to float32 for compatibility
        batch_features = batch_features.float()
        
        y_pred = model(batch_features)
        y_pred = (y_pred>0.8).float()
        
        batch_accuracy = (y_pred.view(-1)== batch_labels).float().mean().item()
        accuracy_list.append(batch_accuracy)
    
overall_accuracy = sum(accuracy_list)/ len(accuracy_list)
print(f'Accuracy: {overall_accuracy:.4f}')
    
```
