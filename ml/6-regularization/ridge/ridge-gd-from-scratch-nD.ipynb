```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor
```

```python
rr = SGDRegressor(penalty='l2', max_iter=500, eta0=0.1, learning_rate='constant' , alpha=0.001)
rr.fit(X_train,y_train)
print(rr.coef_)
print(rr.intercept_)
```

```python
y_pred = rr.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

```python
from sklearn.linear_model import Ridge
```

```python
rr = Ridge(alpha=0.001 , max_iter=500 ,solver='sparse_cg')
rr.fit(X_train,y_train)
print(rr.coef_)
print(rr.intercept_)
y_pred = rr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
class MyRidge:
    
    def __init__(self, epochs, learning_rate, alpha):
        
        self.alpha = alpha
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.coef_ = None
        self.intercept_ = None
        
        
    def fit(self ,X_train, y_train):
        
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        theta = np.insert(self.coef_ , 0 , self.intercept_)
        
        X_train = np.insert(X_train,0,1, axis=1)
        
        for i in range(self.epochs):
            theta_der = np.dot(X_train.T, X_train ).dot(theta) - np.dot(X_train.T , y_train ) + self.alpha*theta
            theta = theta - self.learning_rate*theta_der
        
        self.intercept_ = theta[0]
        self.coef_ = theta[1:]
        print(self.intercept_ , self.coef_)
        
    def predict(self, X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
mridge = MyRidge(epochs=500 , learning_rate=0.005 ,alpha=0.001)
mridge.fit(X_train,y_train)
y_pred = mridge.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
