```python
from ultralytics import YOLO

# Load pretrained classifier
model = YOLO("yolov8n-cls.pt")

# Train
results = model.train(data="/kaggle/input/mosquito-dataset-in-train-test-split", epochs=100, imgsz=224 ,batch=128, workers=2, device=0 )
```

```python
metrics = model.val()
print("Top-1 Accuracy:", metrics.top1)
print("Top-5 Accuracy:", metrics.top5)
```

```python
import random
import os
from PIL import Image
import matplotlib.pyplot as plt

test_dir = "/kaggle/input/mosquito-dataset-in-train-test-split/test"

# Pick random class and image
cls = random.choice(os.listdir(test_dir))
img_path = os.path.join(test_dir, cls, random.choice(os.listdir(os.path.join(test_dir, cls))))

# Load image with PIL (more reliable than mpimg in Kaggle input)
img = Image.open(img_path)

# Predict
results = model.predict(img_path)
res = results[0]

# Get top1 class and confidence
pred_class = res.names[res.probs.top1]
confidence = res.probs.top1conf.item()

# Plot image
plt.figure(figsize=(5,5))
plt.imshow(img)
plt.axis("off")
plt.title(f"Predicted: {pred_class} ({confidence:.2f})\nTrue: {cls}", fontsize=12, color="blue")
plt.show()
```

```python
import random
import os
from PIL import Image
import matplotlib.pyplot as plt

test_dir = "/kaggle/input/mosquito-dataset-in-train-test-split/test"

# Create a 2x3 grid for images
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.flatten()

for ax in axes:
    # Pick a random class and image
    cls = random.choice(os.listdir(test_dir))
    img_path = os.path.join(test_dir, cls, random.choice(os.listdir(os.path.join(test_dir, cls))))

    # Load image
    img = Image.open(img_path)

    # Predict
    results = model.predict(img_path)
    res = results[0]

    # Get top1 prediction
    pred_class = res.names[res.probs.top1]
    confidence = res.probs.top1conf.item()

    # Display
    ax.imshow(img)
    ax.axis("off")
    ax.set_title(f"P: {pred_class} ({confidence:.2f})\nT: {cls}", fontsize=10)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import numpy as np
import random
import os

y_true, y_pred = [], []

for cls in os.listdir(test_dir):
    cls_dir = os.path.join(test_dir, cls)
    for img_name in random.sample(os.listdir(cls_dir), min(20, len(os.listdir(cls_dir)))):  # sample 20 images per class
        img_path = os.path.join(cls_dir, img_name)
        results = model.predict(img_path)
        res = results[0]

        # ✅ Correct way to get predicted class
        pred_class = res.names[res.probs.top1]

        y_true.append(cls)
        y_pred.append(pred_class)

labels = ['AEDES', 'ANOPHELES', 'CULEX']
cm = confusion_matrix(y_true, y_pred, labels=labels)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
disp.plot(cmap="Blues", xticks_rotation=30)
plt.title("Confusion Matrix (sampled)")
plt.show()
```
