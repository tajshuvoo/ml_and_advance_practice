```python
model = Sequential()

model.add(Embedding(10000, output_dim=2))
model.build(input_shape=(None, 50))
model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# ===============================
# TPU SETUP
# ===============================
try:
    tpu = tf.distribute.cluster_resolver.TPUClusterResolver()
    print('Running on TPU:', tpu.master())
except ValueError:
    tpu = None

if tpu:
    tf.config.experimental_connect_to_cluster(tpu)
    tf.tpu.experimental.initialize_tpu_system(tpu)
    strategy = tf.distribute.TPUStrategy(tpu)
else:
    strategy = tf.distribute.get_strategy()  # CPU/GPU fallback

print("Number of replicas:", strategy.num_replicas_in_sync)

# ===============================
# LOAD IMDB DATA
# ===============================
max_features = 20000  # number of words to consider as features
maxlen = 200          # cut texts after this number of words

(x_train, y_train), (x_test, y_test) = keras.datasets.imdb.load_data(num_words=max_features)
x_train = keras.preprocessing.sequence.pad_sequences(x_train, maxlen=maxlen)
x_test = keras.preprocessing.sequence.pad_sequences(x_test, maxlen=maxlen)

# ===============================
# MODEL inside strategy scope
# ===============================
with strategy.scope():
    model = keras.Sequential([
        layers.Embedding(max_features, 128),
        layers.SimpleRNN(64, return_sequences=False),
        layers.Dense(1, activation="sigmoid"),
    ])
    model.build(input_shape=(None, maxlen))

    model.compile(
        optimizer="adam",
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

model.summary()

# ===============================
# TRAINING
# ===============================
history = model.fit(
    x_train, y_train,
    batch_size=128 * strategy.num_replicas_in_sync,  # scale batch size
    epochs=20,
    validation_data=(x_test, y_test)
)
```
